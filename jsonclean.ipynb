```python
import json

with open("metformin.json", "r", encoding="utf-8") as file:
    data = json.load(file)

print("Number of sections:", len(data))
```

```python
import re

def clean_text(text):
    # Replace tabs and newlines with spaces
    text = text.replace("\n", " ")
    text = text.replace("\t", " ")

    # Remove excessive spaces
    text = re.sub(r"\s+", " ", text)

    # Fix spaces before punctuation
    text = re.sub(r"\s+([,.!?;:])", r"\1", text)

    # Fix spaces around brackets
    text = re.sub(r"\(\s+", "(", text)
    text = re.sub(r"\s+\)", ")", text)

    # Remove spaces around hyphens where clearly formatting-related
    text = re.sub(r"\s+-\s+", "-", text)

    return text.strip()
```

```python
cleaned_data = []

for item in data:

    cleaned_item = {
        "drug": item["drug"],
        "section": clean_text(item["section"]),
        "text": clean_text(item["text"])
    }

    cleaned_data.append(cleaned_item)
```

```python
with open("metformin_cleaned.json", "w", encoding="utf-8") as file:
    json.dump(
        cleaned_data,
        file,
        indent=4,
        ensure_ascii=False
    )

print("Cleaned JSON saved successfully!")
```

```python
import re

def clean_text(text):

    # Replace newlines and tabs with spaces
    text = text.replace("\n", " ")
    text = text.replace("\t", " ")

    # Remove excessive whitespace
    text = re.sub(r"\s+", " ", text)

    # Fix formatting around hyphens
    text = re.sub(r"(\w)-\s+(\w)", r"\1-\2", text)

    # Fix common unit formatting
    text = text.replace("m 2", "m²")
    text = text.replace("m 2", "m²")

    # Remove unnecessary spaces before punctuation
    text = re.sub(r"\s+([,.;:!?])", r"\1", text)

    # Clean spaces inside brackets
    text = re.sub(r"\(\s+", "(", text)
    text = re.sub(r"\s+\)", ")", text)

    return text.strip()
```

```python
cleaned_data = []

for item in data:
    cleaned_data.append({
        "drug": item["drug"],
        "section": clean_text(item["section"]),
        "text": clean_text(item["text"])
    })

with open("metformin_cleaned.json", "w", encoding="utf-8") as file:
    json.dump(
        cleaned_data,
        file,
        indent=4,
        ensure_ascii=False
    )

print("Cleaned JSON saved!")
```
