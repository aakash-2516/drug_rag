```python
import json

with open("metformin_cleaned.json", "r", encoding="utf-8") as file:
    data = json.load(file)

print("Sections loaded:", len(data))
```

```python
def create_chunks(text, chunk_size=1500, overlap=200):
    chunks = []

    start = 0

    while start < len(text):

        end = start + chunk_size

        # If we're not at the end, try to end at a sentence
        if end < len(text):
            sentence_end = text.rfind(". ", start, end)

            if sentence_end > start:
                end = sentence_end + 1

        chunk = text[start:end].strip()

        if chunk:
            chunks.append(chunk)

        # Move forward while keeping overlap
        start = end - overlap

    return chunks
```

```python
chunked_data = []

for item in data:

    chunks = create_chunks(item["text"])

    for i, chunk in enumerate(chunks):

        chunked_data.append({
            "chunk_id": f"{item['drug']}_{i+1}",
            "drug": item["drug"],
            "section": item["section"],
            "text": chunk
        })

print("Total chunks:", len(chunked_data))
```

```python
with open("metformin_chunks.json", "w", encoding="utf-8") as file:
    json.dump(
        chunked_data,
        file,
        indent=4,
        ensure_ascii=False
    )

print("Chunks saved successfully!")
```
